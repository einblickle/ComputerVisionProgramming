# Arbeiten mit Bildern

## Import der Bibliotheken

```python
import numpy as np
import scipy
import cv2
import matplotlib.pyplot as plt
import plotly
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

### Laden eines Testbildes

```python
image_baboon=cv2.imread('./Images/baboon.png', cv2.IMREAD_COLOR)
# image taklen from https://github.com/scijs/baboon-image/blob/master/baboon.png
# image is loaded as BRG image. convert to RGB
image_baboon = cv2.cvtColor(image_baboon, cv2.COLOR_BGR2RGB)
```

#### Anzeigen der Datenstruktur des Bildes
Dieses Farb - Bild wird als numpy ndarray, einer v x u x 3 Zahlen - Matrix geladen.

Die Koordinaten u, v repraesentieren hier die Pixelposition im Bild, die 3. Koordinate die jeweiligen Farbkanäle Rot, Gruen, und Blau (rgb)

Bildgröße und Datentyp können durch folgende Befehle abgefragt werden.
 

```python
dataType = image_baboon.dtype
imageShape = image_baboon.shape
print(f'Datentyp des Bildes: {dataType}')
print(f'fortmat des Bildes: {imageShape}')
```

### Anzeigen des Bildes 

#### Inline Anzeige des Bildes
Bewegen den Mauszeige über das Bild, dann können Sie sowohl die Werte der Farbkanäle als auch die dazugehörigen Pixelkoordinaten auslesen.

```python
# Anzeige des Bildes
height, width = image_baboon.shape[:2]
fig = px.imshow(image_baboon)
fig.update_layout(width=width, height=height + 50, margin=dict(l=0, r=0, t=50, b=0))
fig.show()
```

### Farbkanäle einzeln Anzeigen

Die Farbkanaele des Bildes lassen sich einzeln ansteuern

Roter Kanal `image_baboon[:,:,0]`

Grüner Kanal `image_baboon[:,:,1]`

Blauer Kanal `image_baboon[:,:,2]`

```python
image_baboon_red = image_baboon[:,:,0]
image_baboon_green = image_baboon[:,:,1]
image_baboon_blue = image_baboon[:,:,2]

## Plot the images
height, width = image_baboon.shape[:2]

# Anzeige des roten Kanals
fig = px.imshow(image_baboon_red, color_continuous_scale='gray', title='Roter Farbkanal')
fig.update_layout(width=width, height=height + 50, margin=dict(l=0, r=0, t=50, b=0))
fig.show()

# Anzeige des grünen Kanals
fig = px.imshow(image_baboon_green, color_continuous_scale='gray', title='Gruener Farbkanal')
fig.update_layout(width=width, height=height + 50, margin=dict(l=0, r=0, t=50, b=0))
fig.show()

# Anzeige des blauen Kanals
fig = px.imshow(image_baboon_blue, color_continuous_scale='gray', title='Blauer Farbkanal')
fig.update_layout(width=width, height=height + 50, margin=dict(l=0, r=0, t=50, b=0))
fig.show()
```

### Filterung nach einer Speziellen Farbe / Farbbereich

mit der Funktion cv2.inRange lässt sich das Bild nach einem bestimmten Farbbereich Filtern. Die Funktion cv2.inRange(image, lowerb, upperb)
Dabei werden die Pixel herausgefiltert deren Farbwert (r, g, b) zwischen der unteren Farbgrenze (lowerb) und der oberen Farbgrenze (upperb) liegt. Im Nachfolgenden Beispiel ist:
`lowerb = (200, 0, 0)` und `upperb = (255, 100, 100)`
Somit Filtert diese Funktion Farbwerte: 
200 < r <= 255 und 0 < g < 100 und 0 < b < 100)

#### Nase

```python
nose = cv2.inRange(image_baboon, (150, 0, 0), (255, 80, 80))

# Anzeige der Nase
height, width = nose.shape[:2]
fig = px.imshow(nose, color_continuous_scale='gray', title='Maskierte Nase')
fig.update_layout(width=width, height=height + 50, margin=dict(l=0, r=0, t=50, b=0))
fig.show()

```

#### Backe

Nachdem eben die Nase des Affen gefiltert wurde soll nun die Backe gefiltert werden. Versuchen Sie den Filter so anzupassen dass nur die zur türkis-blauen Backe gehörenden Pixel detektiert werden. Nutzen Sie hierfür das oben stehende Farbbild und lesen Sie die einzelnen Farbwerte mit dem Mauszeiger ab.

**Achtung** *Sie werden feststellen dass dies nur sehr schwer umsetzbar ist, die orangenen Augen lassen sich nur schwer von der roten Nase trennen*

```python
## Verändern Sie die Filterwerte durch Anpassen der unteren Schranke und der oberen Schranke
jaw = cv2.inRange(image_baboon, (0, 0, 0), (250, 255, 255)) 

# Anzeige der Nase
height, width = jaw.shape[:2]
fig = px.imshow(jaw, color_continuous_scale='gray', title='Maskierte Nase')
fig.update_layout(width=width, height=height + 50, margin=dict(l=0, r=0, t=50, b=0))
fig.show()
```

#### Wechsel des Farbraums

Der RGB-Farbraum entspricht nicht unbedingt der menschlichen Farbwahrnehmung, in dem Sinne, dass ähnliche Farben (r,g,b) auch zahlenmäßig nahe beieinanderliegen. Der HSV Farbraum parametrisiert die Farben so, dass ähnliche Farben jetzt auch örtlich nebeneinander liegen.

Das Bild lässt sich nun deutlich einfacher nach dem Farbtonwert filtern (hue)

```python
# konvertiere RGB zu HSV
image_baboon_hsv = cv2.cvtColor(image_baboon, cv2.COLOR_RGB2HSV)

## Verändern Sie die Filterwerte durch Anpassen der unteren Schranke und der oberen Schranke
jaw = cv2.inRange(image_baboon_hsv, (14, 215, 60), (25, 255, 255))



# Anzeige des Bildes
# Achtung: Plotly kann im HSV-Raum abgespeicherte Bilder nicht darstellen. Das Dargestellte Bild ist somit ein Falschfarbenbild
height, width = image_baboon_hsv.shape[:2]
fig = px.imshow(image_baboon_hsv, color_continuous_scale='hsv', title='Bild im HSV Raum')
fig.update_layout(width=width, height=height + 50, margin=dict(l=0, r=0, t=50, b=0))
fig.show()

# Anzeige der Nase
height, width = jaw.shape[:2]
fig = px.imshow(jaw, color_continuous_scale='gray', title='Bild im HSV Raum')
fig.update_layout(width=width, height=height + 50, margin=dict(l=0, r=0, t=50, b=0))
fig.show()



```

### Erkennung von Verkehrszeichen

```python
image_verkehr=cv2.imread('./Images/verkehr_001_c.jpg', cv2.IMREAD_COLOR)
# image taklen from https://github.com/scijs/baboon-image/blob/master/baboon.png
# image is loaded as BRG image. convert to RGB
image_verkehr = cv2.cvtColor(image_verkehr, cv2.COLOR_BGR2RGB)

# konvertiere RGB zu HSV
image_verkehr_hsv = cv2.cvtColor(image_verkehr, cv2.COLOR_RGB2HSV)
```

```python
# Anzeige des Bildes
# Achtung: Plotly kann im HSV-Raum abgespeicherte Bilder nicht darstellen. Das Dargestellte Bild ist somit ein Falschfarbenbild
fig = px.imshow(image_verkehr, color_continuous_scale='hsv', title='Bild im RGB Raum')
fig.update_layout(width = 500, height=400, margin=dict(l=0, r=0, t=50, b=0))
fig.show()

# Anzeige der Nase
fig = px.imshow(image_verkehr_hsv, color_continuous_scale='gray', title='Bild im HSV Raum')
fig.update_layout(width = 500, height=400, margin=dict(l=0, r=0, t=50, b=0))
fig.show()
```

```python
## Verändern Sie die Filterwerte durch Anpassen der unteren Schranke und der oberen Schranke
filter_vorfahrt = cv2.inRange(image_verkehr_hsv, (16, 170, 50), (22, 255, 255))
filter_zebra = cv2.inRange(image_verkehr_hsv, (100, 180, 50), (110, 255, 255))
filter_tempo = cv2.inRange(image_verkehr_hsv, (170, 100, 50), (178, 255, 254))


fig = px.imshow(filter_vorfahrt, color_continuous_scale='gray', title='Bild im RGB Raum')
fig.update_layout(width = 500, height=400, margin=dict(l=0, r=0, t=50, b=0))
fig.show()

fig = px.imshow(filter_zebra, color_continuous_scale='gray', title='Bild im RGB Raum')
fig.update_layout(width = 500, height=400, margin=dict(l=0, r=0, t=50, b=0))
fig.show()

fig = px.imshow(filter_tempo, color_continuous_scale='gray', title='Bild im RGB Raum')
fig.update_layout(width = 500, height=400, margin=dict(l=0, r=0, t=50, b=0))
fig.show()
```
